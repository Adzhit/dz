```python
import numpy as np

A = np.array([
    [ 17.16,  18.40,  20.47,  -0.60],
    [-24.74, -23.83, -19.20,   8.43],
    [ 25.95,  -6.09,  -9.09, -11.47],
    [ -0.23,   9.64, -10.32, -16.75]
])
B = np.array([-17.19, 27.62, 2.37, 25.55])

X = np.linalg.solve(A, B)
print("Решение линейной системы (Задание 4.1):")
for i, x in enumerate(X):
    print(f"x{i+1} = {x:.6f}")
```

```python
import numpy as np
from scipy.optimize import fsolve

def equations(vars):
    x, y = vars
    eq1 = x**2 * y**2 - ((x**2 - y**2)**2) / 4 - 0.5 - x
    eq2 = x * y * (y**2 - x**2) + 0.5 - y
    return [eq1, eq2]

# Используем [0,0], так как графики не пересекаются (действительных корней нет)
solution = fsolve(equations, [0.0, 0.0])
print("\nПопытка численного решения нелинейной системы (Задание 4.2):")
print(f"x = {solution[0]:.6f}, y = {solution[1]:.6f}")
print("Примечание: алгоритм не сойдется к правильному ответу, так как действительных корней нет.")
```
