```python
%%writefile lab_1_1.py
import math

def divide(numerator, denominator):
    if denominator == 0:
        if numerator == 0:
            raise ZeroDivisionError("Ділення на нуль не підтримується (чисельник і знаменник нульові)")
        raise ZeroDivisionError("Ділення на нуль не підтримується")
    return numerator / denominator

def calculate_variant_1(x, y, z, b):
    numerator_a = 2 * math.cos(x - math.pi / 6) * b
    denominator_a = 0.5 + math.sin(y)**2
    a = divide(numerator_a, denominator_a)
    
    # Используем abs(y) для избежания комплексных чисел при дробной степени отрицательного y
    denominator_m_raw = 3 + abs(y)**(2/5)
    m = 1 + divide(z**2, denominator_m_raw)
    
    return a, m
```

```python
%%writefile test_lab_1_1.py
import pytest
import math
from lab_1_1 import divide, calculate_variant_1

def test_divide_normal():
    assert divide(10, 2) == 5.0

def test_divide_by_zero():
    with pytest.raises(ZeroDivisionError, match="Ділення на нуль не підтримується"):
        divide(10, 0)

def test_divide_both_zeros():
    with pytest.raises(ZeroDivisionError):
        divide(0, 0)

def test_calculate_variant_1_approximation():
    x, y, z, b = 1.45, -1.22, 3.5, 3.07
    a, m = calculate_variant_1(x, y, z, b)

    # Подставьте фактические значения, которые у вас получились при выводе: a ≈ 2.66905676, m ≈ 4.35375523
    assert a == pytest.approx(2.66905676, abs=1e-8)
    assert m == pytest.approx(4.35375523, abs=1e-8)
```

Відповідь: Стандартні граничні значення (наприклад, нуль, одиниця або екстремальні точки поза початковим діапазоном), які не охоплюються автоматично згенерованими наборами даних, оскільки вони вимагають ручного додавання окремих тестових сценаріїв для перевірки особливих умов чи винятків.

2)Звідки взяті допуски і в який момент застосовується допуск до виходу?

Відповідь: Допуски (abs=1e-7 / abs=1e-8) визначаються виходячи з необхідної точності обчислень лабораторної роботи та можливої похибки округлення чисел із плавною комою (float) у Python. Допуск застосовується під час перевірки результатів у тестах за допомогою функції pytest.approx на етапі порівняння очікуваного та фактичного значень.

3)Чому перехід на більш високий тип даних не гарантує достовірність розумного відповіді?

Відповідь: Перехід на типи з більшою розрядністю (наприклад, з float на Decimal чи double) не усуває математичну похибку самої моделі чи початкових похибок вхідних даних, якщо самі параметри містять невизначеності або наближаються до критичних точок (наприклад, знаменник, що прямує до нуля).
