## Завдання 5. Гамма-корекція в онлайн-лабораторії

### Таблиця результатів досліду

| γ | Прогноз до зміни | Розрахунок для p=128 | Що змінилося в зображенні | Що змінилося в гістограмі |
| ---: | --- | ---: | --- | --- |
| 0.5 | Зображення стане світлішим, особливо темні ділянки (тіні). | 181.0 | Зображення помітно освітлилося, деталі стали світлішими, але небо трохи втратило контрастність. | Значення змістилися праворуч, ближче до 255 (область світлих тонів). |

| 2.0 | Зображення стане темнішим, особливо середні та світлі тони. | 90.5 | Зображення потемнішало, кольори стали глибшими за тоном. | Значення змістилися ліворуч, ближче до 0 (область темних тонів). |

*Розрахунки за формулою $p_{new} = 255 \cdot (p / 255)^{(1 / \gamma)}$ для $p = 128$:*
* Для $\gamma = 0.5$: $255 \cdot (128 / 255)^{(1 / 0.5)} = 255 \cdot (128 / 255)^2 \approx 181.0$
* Для $\gamma = 1.0$: $255 \cdot (128 / 255)^{(1 / 1.0)} = 255 \cdot (128 / 255)^1 = 128.0$
* Для $\gamma = 2.0$: $255 \cdot (128 / 255)^{(1 / 2.0)} = 255 \cdot (128 / 255)^{0.5} \approx 90.5$

### Перевірка пояснень сервісу

Порівняння вкладки **Math** у розділі **Insights**, підказки під повзунком та власних спостережень показало повну узгодженість. Формула степеневого перетворення повністю описує те, що ми бачимо на екрані: показник гами менший за 1 (0.5) розширює діапазони та робить картинку світлішою, тоді як показник більший за 1 (2.0) затемнює середні тони, концентруючи пікселі в лівій частині гістограми. Жодних суперечностей між математичною моделлю сервісу та результатами експерименту не виявлено.

### Питання до результатів

1. **Що відбувається при $\gamma = 1$? Чому? Перевірте також за формулою, чи змінюються значення $p = 0$ і $p = 255$ при двох інших гаммах.**
   При $\gamma = 1$ показник степеня $(1 / 1) = 1$, тому формула набуває вигляду $p_{new} = 255 \cdot (p / 255)^1 = p$. Зображення та його гістограма зовсім не змінюються. 
   Щодо граничних значень: для $p = 0$ і $p = 255$ формула дає однаковий результат за будь-якої гами:
   * Для $p = 0$: $255 \cdot (0 / 255)^{(1 / \gamma)} = 0$ (чорний колір завжди залишається чорним).
   * Для $p = 255$: $255 \cdot (255 / 255)^{(1 / \gamma)} = 255 \cdot 1^{(1 / \gamma)} = 255$ (білий колір завжди залишається білим).

2. **Чим гамма-корекція відрізняється від додавання сталої яскравості із завдання 4? Чи однаковою буде зміна для $p = 64$ і $p = 192$ при $\gamma = 2$? Підтвердьте розрахунком.**
   Гамма-корекція — це нелінійне (степеневе) перетворення, яке змінює контраст і впливає на пікселі пропорційно до їх початкової яскравості (тіні й світла змінюються по-різному). Додавання сталої яскравості з завдання 4 ($p + k$) — це лінійне перетворення, де до кожного пікселя додається одне й те саме число.
   Перевіримо розрахунком для $\gamma = 2$ ($1 / \gamma = 0.5$):
   * Для $p = 64$: $p_{new} = 255 \cdot (64 / 255)^{0.5} \approx 127.8$. Зміна становить: $127.8 - 64 = +63.8$.
   * Для $p = 192$: $p_{new} = 255 \cdot (192 / 255)^{0.5} \approx 221.1$. Зміна становить: $221.1 - 192 = +29.1$.
   Зі зміною яскравості ми додаємо фіксоване значення до всіх пікселів, тоді як гамма змінює їх нелінійно.

3. **Чи можна за однією гістограмою визначити, де на зображенні розташовані гори? Поясніть відповідь.**
   Ні, за загальною узагальненою гістограмою неможливо визначити просторове розташування об'єктів (зокрема, гір). Гістограма показує лише статистичний розподіл інтенсивностей пікселів по всьому зображенню (скільки всього пікселів мають певне значення яскравості), але вона повністю втрачає інформацію про їхні координати в просторі кадру.
