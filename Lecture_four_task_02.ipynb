Питання до результату: чому фраза «на екрані один корінь» ще не означає «у рівняння лише один корінь»? **Відповідь:** Кількість видимих коренів залежить від поточного масштабу та меж вікна перегляду графіка. При зміщенні області або зміні масштабу можуть з'явитися інші корені рівняння, які не поміщалися в початковий екран.
